```python
import matplotlib.pyplot as plt
```

```python
X = [1, 2, 3, 4]
Y = [5, 6, 7, 8]

plt.plot(X, Y)
plt.show()
```

```python
# Plot methods
movies = ["The Dark Knight", "The Hurt Locker", "The King's Speech", "The Artist", "Argo"]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232]

plt.plot(years, revenue)
plt.title("Oscar movies revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# Merge plot
movies2 = ["Slumdog Millionaire", "Avatar", "Inception", "Hugo", "Lincoln"]
revenue2 = [378, 2788, 829, 185, 275]

plt.plot(years, revenue, label="Oscar movies")
plt.plot(years, revenue2, label="Non-Oscar movies")
plt.title("Oscar Vs Non-Oscar movie revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend(loc="upper right")
```

```python
# Format Strings
plt.plot(years, revenue, "ro--")  # red circle dash
plt.title("Oscar Movie Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
```

```python
# Styling
plt.plot(movies, revenue)
plt.title("Oscar Movie Revenue")
plt.xlabel("Movie name")
plt.ylabel("Revenue (in $M)")

# plt.style.available     # list down all available themes
plt.style.use("default")  # defines theme of plot
plt.grid()                # creates grid on background
plt.tight_layout()        # prevent overlapping of labels
plt.show()
plt.savefig("final_plot.png")  # download
```
