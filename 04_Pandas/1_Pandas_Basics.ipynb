```python
import pandas as pd
```

### Series

```python
# Create
s = pd.Series([1,2,3,4,5])
print(s, type(s))
print("Value at index 1: ",s[1])

s = pd.Series([23,24,25,26], index=["Aman","Pranjal","Ayush","Kaustuv"])
print(s)
print("Age of aman: ",s["Aman"])
print("Age of ayush: ", s.iloc[2])
print(s.index)
```

```python
# Operations
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])
print(s1+s2)

s1[0] = 2  # update element
print(s1)

s = s1.drop(0) # drop element
print(s)
```

### Dataframe
- Note: Everytime dataframe will be created as view. To create copy, use .copy()

```python
# Create DataFrame
info = {
    "Name": ["Aman", "Ayush", "Kaustuv", "Pranjal"],
    "Age": [25, 23, 24, 22],
    "CGPA": [9.8, 8.4, 9.2, 8.8]
}
# Using dictionary
df = pd.DataFrame(info)  
print(df, "\n",type(df), "\n")

# Using 2d array
df = pd.DataFrame([["Aman", 25], ["Ayush", 23], ["Kaustuv", 24], ["Pranjal", 22]], columns=["Name", "Age"])
print(df, "\n")

# Using numpy array
import numpy as np
np_arr = np.array([[1,2,3],[4,5,6]])
df = pd.DataFrame(np_arr, columns=["A", "B", "C"])
print(df, "\n")

print(df.columns)        # Columns Headers
print(df.index)          # Row Headers
```

```python
# Reading from a file
df = pd.read_csv("employee_data.csv")   # csv
print(df, type(df), "\n")

df = pd.read_json("employee_data.json") # json
print(df, type(df))
```

```python
# Methods
print(df.head(), "\n")     # top 5
print(df.tail(), "\n")     # last 5
print(df.sample(3), "\n")  # random
print(df.info(), "\n")     # information
print(df.shape, "\n")      # shape
print(df.describe(), "\n") # description of numerical columns
print(df.columns, "\n")    # list of columns
print(df.nunique(), "\n")  # no. of unique rows per columns
```

### Practice on AQI Dataset

```python
# Read csv
df = pd.read_csv("globalAirQuality.csv")
print(df.sample(5))
```

```python
# Selecting columns
df[["city", "aqi"]].head()
```

```python
# Selecting rows
df.loc[2]
df.loc[0:2]  # start idx : end idx(inclusive)

df.iloc[2]
df.iloc[0:2] # start idx : end idx(exclusive)
```

```python
# Selecting rows & columns
df.loc[0:2, ["city", "aqi"]]

df.iloc[0:2, [2,3,4,11]]
df.iloc[0:2, 1:5]
```

```python
# Selecting cell
df.at[1, "city"]

df.iat[1, 2]
```

```python
# Filtering Data
filtered_df = df[(df["aqi"] > 100) & (df["temperature"] > 30)][["city","aqi","temperature"]]
filtered_df
```

```python
# Differenc between label(original) & index
print("Label 6 data:\n", filtered_df.loc[6])
print("\nIndex 0 data:\n", filtered_df.iloc[0])
```

```python
# Filter using query
df.query("aqi < 100 & temperature < 30")[["city", "aqi", "temperature"]]
```
