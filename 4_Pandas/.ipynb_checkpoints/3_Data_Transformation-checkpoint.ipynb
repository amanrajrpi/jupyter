```python
import pandas as pd
```

### Feature Engineering

```python
df = pd.read_csv("raw_data.csv").copy()
df.head(3)
```

```python
# apply(fn)
df["tax"] = df["income"].apply(lambda x : "20%" if x >= 60000 else "10%")
df.head(3)
```

```python
# map()
gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}
df["gender"] = df["gender"].map(gender_map)
df.head(3)
```

```python
# assign()
df = df.assign(new_income = df["income"] * 1.1)
df.head(3)
```

```python
# replace(old,new)
df["country"] = df["country"].replace("USA", "US")
df.head(3)
```
