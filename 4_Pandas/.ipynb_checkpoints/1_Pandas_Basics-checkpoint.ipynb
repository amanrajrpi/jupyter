```python
import pandas as pd
```

### Series

```python
# Create
s = pd.Series([1,2,3,4,5])
print(s, type(s))
print("Value at index 1: ",s[1])

s = pd.Series([23,24,25,26], index=["Aman","Pranjal","Ayush","Kaustuv"])
print(s)
print("Age of aman: ",s["Aman"])
print("Age of ayush: ", s.iloc[2])
print(s.index)
```

```python
# Operations
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])
print(s1+s2)

s1[0] = 2  # update element
print(s1)

s = s1.drop(0) # drop element
print(s)
```

### Dataframe

```python
# Create DataFrame
info = {
    "Name": ["Aman", "Ayush", "Kaustuv", "Pranjal"],
    "Age": [25, 23, 24, 22],
    "CGPA": [9.8, 8.4, 9.2, 8.8]
}
# Using dictionary
df = pd.DataFrame(info)  
print(df, "\n",type(df), "\n")

# Using 2d array
df = pd.DataFrame([["Aman", 25], ["Ayush", 23], ["Kaustuv", 24], ["Pranjal", 22]], columns=["Name", "Age"])
print(df, "\n")

# Using numpy array
import numpy as np
np_arr = np.array([[1,2,3],[4,5,6]])
df = pd.DataFrame(np_arr, columns=["A", "B", "C"])
print(df, "\n")

print(df.columns)        # Columns Headers
print(df.index)          # Row Headers
```
