```python
import pandas as pd
```

### Merging & Joining Data

```python
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})
print(df_customers)

df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
print("\n", df_orders)
```

```python
# Inner Join
pd.merge(df_customers, df_orders, on="customer_id")
```

```python
# Left Join
pd.merge(df_customers, df_orders, on="customer_id", how="left")
```

```python
# Right Join
pd.merge(df_customers, df_orders, on="customer_id", how="right")
```

```python
# Outer Join
pd.merge(df_customers, df_orders, on="customer_id", how="outer")
```

### Concatination

```python
df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "name": ["Adam", "Eve", "Bob"]
})
print(df1)

df2 = pd.DataFrame({
    "id": [4, 5, 6],
    "name": ["Charlie", "Fester", "Gilian"]
})
print("\n", df2)
```

```python
# Vertically (row-wise)
pd.concat([df1, df2], ignore_index=True)
```

```python
# Horizontally (column-wise)
pd.concat([df1,df2], axis=1)
```
