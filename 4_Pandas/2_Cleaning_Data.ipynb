```python
import pandas as pd
```

```python
df = pd.read_csv("raw_data.csv")
df.head()
```

### Handle Missing Values

```python
# Check Null values
print(df.isnull())             # check null values
print("\n", df.isnull().sum()) # count null values
```

```python
# Drop Null values
print(df.dropna())       # drop rows
print(df.dropna(axis=1)) # drop columns
```

```python
# Replace Null values
print(df.fillna(0))        # replace all null values
mean_age = df["age"].mean()
print(df["age"].fillna(mean_age)) # replace null values for a column
```

```python
# Forward Fill - Fill top neighbor value
print(df.ffill())
```

```python
# Backward Fill - Fill bottom neighbor value
print(df.bfill())
```

### Handle Duplicate Values

```python
# Check duplicate values
print(df.duplicated())
print("\n", df["country"].duplicated())
```

```python
# Drop duplicates
df1 = df.copy()
df1.drop_duplicates(inplace=True)  # Inplace is used to save changes
df1
```

### Handle DataTypes & date-time

```python
# Check datatypes
df.dtypes
```

```python
# Change datatypes
df1 = df.copy()
df1 = df1.fillna(0)
df1["age"] = df1["age"].astype("int64")
df1.dtypes
```

```python
# Change Date-Time
df2 = pd.read_csv("globalAirQuality.csv")[["timestamp", "city", "aqi"]]
print(df2.head(3))
print(df2["timestamp"].dtype)

df2["timestamp"] = pd.to_datetime(df2["timestamp"])
print(df2["timestamp"].dtype)
```

### Handle Strings

```python
# Upper, Lower, Capitalize
print(df["country"].str.upper())
print("\n", df["country"].str.lower())
print("\n", df["country"].str.capitalize())
```

```python
# Split
df["name"].str.split(" ")
```

```python
# Strip
df["name"].str.strip()
```

```python
# Contains
df["country"].str.contains("india", case=False)
```
