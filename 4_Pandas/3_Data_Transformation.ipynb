```python
import pandas as pd
```

### Feature Engineering

```python
df = pd.read_csv("raw_data.csv").copy()
df.head(3)
```

```python
# apply(fn)
df["tax"] = df["income"].apply(lambda x : "20%" if x >= 60000 else "10%")
df.head(3)
```

```python
# map()
gender_map = {"Male" : "M", "Female" : "F", "Unknown" : "U"}
df["gender"] = df["gender"].map(gender_map)
df.head(3)
```

```python
# assign()
df = df.assign(new_income = df["income"] * 1.1)
df.head(3)
```

```python
# replace(old,new)
df["country"] = df["country"].replace("USA", "US")
df.head(3)
```

```python
# rename( columns={old:new} )
df = df.rename(columns = {"income":"salary"})  # rename column
print(df.head(3))

df = df.rename(index = {1:"A"})                # rename row index
print("\n", df.head(3))

df.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax", "New_Income"]  # rename entire columns
print("\n", df.head(3))
```

```python
# Sort
sorted_df = df.sort_values("Income")          # Sort on single column
print(sorted_df.head(3))

sorted_df = df.sort_values(["Income", "Age"]) # Sort on multiple columns
print("\n", sorted_df.head(3))

reset_df = sorted_df.reset_index()            # Create new index while preserving old
print("\n", reset_df.head(3))

reset_df = sorted_df.reset_index(drop=True)   # Create new index
print("\n", reset_df.head(3))
```

```python
# Rank
sorted_df["Ranking"] = sorted_df["Income"].rank()  # assign rank to each new value, if tie then avg is assigned
print(sorted_df)

sorted_df["Ranking"] = sorted_df["Income"].rank(ascending=False, method="dense")  # if tie then same rank
print("\n", sorted_df)
```

```python
# Reordering columns
df = df[["Id", "Name", "Age", "Country", "Gender", "Tax", "Income", "New_Income"]]
df.head(3)
```

### Grouping & Aggregation

```python
df.groupby("Country")["Income"].count()  # Count
df.groupby("Country")["Income"].min()    # Min
df.groupby("Country")["Income"].max()    # Max
df.groupby("Country")["Income"].mean()   # Mean
```

```python
# aggregate()
df.groupby("Country")["Income"].agg(["count", "min", "max", "mean"])
df.groupby("Country")["Income"].agg(total_count = "count", min_salary = "min", max_salary = "max", avg_salary = "mean")  # rename columns
```

```python
# aggregate on multiple columns
df.groupby("Country").agg({
    "Age" : "mean",
    "Income" : "max"
})

df.groupby("Country").agg(
    avg_age = ("Age", "mean"),
    max_salary = ("Income", "max")
)
```
