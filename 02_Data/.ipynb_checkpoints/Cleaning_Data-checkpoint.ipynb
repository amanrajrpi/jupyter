```python
import json
```

```python
# load data
def load_data(file):
    with open(file, "r") as f:
        data = json.load(f)
    return data
```

```python
# Print data
data = load_data("customer.json")
for line in data:
    print(line)
```

```python
# Clean & Structure data
def clean_data(data):
    text_to_num = {"one":1, "two":2, "three":3, "four":4, "five":5}
    cleaned_data = []
    unique_users = set()
    for user in data:
        # clean ratings
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user["rating"] = raw_rating
        
        # handle missing values
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        # handle deduplication
        if(user["name"].strip() in unique_users):
            continue
        else:
            unique_users.add(user["name"].strip())
            cleaned_data.append(user)
    return cleaned_data
```

```python
data = clean_data(data)
for line in data:
    print(line)
```

### Data Insights

```python
# avg rating
def get_avg_rating(data):
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])
    return total_rating/len(data)

get_avg_rating(data)
```

```python
# %age of users with poor rating
def poor_rating_percent(data):
    poor_rating = 0
    for user in data:
        if(float(user["rating"])<3):
            poor_rating += 1
    return (poor_rating*100)/len(data)

poor_rating_percent(data)
```
