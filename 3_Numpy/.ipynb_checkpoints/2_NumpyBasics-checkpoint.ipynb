```python
import numpy as np
```

```python
# Create numpy array
arr1 = np.array([1,2,3,4,5])
print(arr1, type(arr1), arr1.dtype, arr1.shape)  # 1-D array

arr2 = np.array([1,2,3,4,5,"aman"])
print(arr2, type(arr2), arr2.dtype, arr2.shape)  # Unicode array

arr3 = np.array([[1,2,3],[4,5,6]])
print(arr3, type(arr3), arr3.dtype, arr3.shape)  # 2-D array

arr4 = np.zeros((2,3))
print(arr4, type(arr4), arr4.dtype, arr4.shape)  # Zero array

arr5 = np.ones((2,3), dtype="int32")
print(arr5, type(arr5), arr5.dtype, arr5.shape)  # Ones array

arr6 = np.full((2,3), 100)
print(arr6, type(arr6), arr6.dtype, arr6.shape)  # Full array

arr7 = np.eye(3)
print(arr7, type(arr7), arr7.dtype, arr7.shape)  # Identity array

arr8 = np.arange(1,11,2)
print(arr8, type(arr8), arr8.dtype, arr8.shape)  # Range elements

arr9 = np.linspace(0, 100, 5)
print(arr9, type(arr9), arr9.dtype, arr9.shape)  # Evenly spaced
```
