```python
import numpy as np
import time
import sys
```

```python
# Performance Test
size = 10_000_000  # 10-million

py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Python list sq time: {end-start} seconds")

np_arr = np.array(py_list)
start = time.time()
sq_arr = np_arr**2
end = time.time()
print(f"Numpy arr sq time: {end-start} seconds")
```

```python
# Memory Test
print(f"Python list size = {sys.getsizeof(py_list)*len(py_list)} bytes")
print(f"Numpy array size = {np_arr.nbytes} bytes")
```
