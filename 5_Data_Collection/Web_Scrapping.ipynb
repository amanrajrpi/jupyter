### Download webpage

```python
import requests

URL = "https://www.scrapethissite.com/pages/simple/"
res = requests.get(URL)

with open("scrapped_data/data.html", "w") as f:
    f.write(res.text)
```

### Parse webpage

```python
from bs4 import BeautifulSoup

with open("scrapped_data/data.html", "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

### Read webpage

```python
# Basic Extration
soup.title.text  # First text
soup.h1.text     # First h1 tag
soup.p.text      # First paragraph
```

```python
# Find Element
soup.find("div")                    # First div
soup.find_all("a")                  # All link
soup.find("li", id="nav_login")     # By Id
soup.find("div", class_="country")  # By class
```

```python
# CSS Selectors
soup.select("h3.country-name")    # Class
soup.select("#nav-login")         # Id
soup.select("div > p")            # Direct child
soup.select("a[href]")            # Attribute
```

```python
# Get Text/Attributes
tag = soup.find("a")

print(tag.text)           # inner text
print(tag.get("href"))    # attribute
print(tag["href"])        # alternative
```

```python
# Loop through Elements
import pandas as pd

data = []
cards = soup.find_all("div", class_="country")

for card in cards:
    country = card.find("h3").get_text(strip=True)
    capital = card.find("span", class_="country-capital").get_text(strip=True)
    population = card.find("span", class_="country-population").get_text(strip=True)
    area = card.find("span", class_="country-area").get_text(strip=True)

    data.append({
        "Country": country,
        "Capital": capital if capital else "N/A",
        "Population": population if population else "N/A",
        "Area": area if area else "N/A"
    })

df = pd.DataFrame(data)
df.head()
```

```python
df.to_csv("scrapped_data/countries.csv", index=False)
```
