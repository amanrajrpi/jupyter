```python
import random
import matplotlib.pyplot as plt

# Generate 100 random numbers between 0 and 100
numbers = [random.randint(0, 100) for _ in range(1000)]

# Plot histogram
plt.hist(numbers, bins=10)

# Show graph
plt.title("Normal Distribution")
plt.xlabel("Numbers")
plt.ylabel("Frequency")

plt.show()
```

```python
import random
import matplotlib.pyplot as plt

# Generate 100 numbers with Gaussian distribution
numbers = [random.gauss(50, 15) for _ in range(1000)]

# Plot histogram
plt.hist(numbers, bins=10)

plt.title("Gaussian Distribution")
plt.xlabel("Numbers")
plt.ylabel("Frequency")

plt.show()
```
